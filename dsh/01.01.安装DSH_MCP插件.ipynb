# 客户端方式

## 安装deepseek harness桌面版

从这里下载安装即可 https://www.deepseek.com/harness/

安装一下命令行工具, 并校验

```
dsh --version
0.2.2-rc.5
```


![](./imgs/dsh_command.png)

1. 点击桌面版左下角的`设置->内置插件->搜索dsh-mcp-center`
